# Germline Variant Calling Exploration & VCF Quality Control (`HG00098`, Chr22)

This notebook performs exploratory and statistical quality control (QC) on `data/raw/sample.vcf` (46,065 variants on human Chromosome 22 from the 1000 Genomes Project sample `HG00098`), evaluating:
1. **Variant Class Breakdown** (Biallelic SNPs vs. Multi-allelic/INDELs and `dbSNP` rsID concordance).
2. **Transition / Transversion (`Ti/Tv`) Ratio** and the 12-class single-nucleotide substitution spectrum.
3. **Genomic Variant Density** across 1-Mb windows on Chromosome 22 (`22q11.1–22q13.33`).
4. **Sample Zygosity (`Het/Hom` Ratio)** across phased genotypes (`0|1`, `1|0`, `1|1`).

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, str(Path("../scripts").resolve()))
from read_vcf import iter_vcf_records
from titv_ratio import compute_titv_metrics

sns.set_theme(style="whitegrid")
```

```python
vcf_path = Path("../data/raw/sample.vcf")
records = list(iter_vcf_records(vcf_path))
titv = compute_titv_metrics(vcf_path)
known = sum(1 for r in records if r.is_known_dbsnp)
print(f"Total variants: {len(records):,} | Known dbSNP: {known:,} ({100*known/len(records):.2f}%) | Novel: {len(records)-known:,} ({100*(len(records)-known)/len(records):.2f}%)")
print(f"Transitions: {titv['transitions']:,} | Transversions: {titv['transversions']:,} | Ti/Tv Ratio: {titv['titv_ratio']:.4f}")
```
